```python
#funcion para explorar las tablas
def explorar(df, nombre):
    print(f"=== {nombre} ===")
    print("Filas y columnas:", df.shape)
    print("\nTipo de cada columna:")
    print(df.dtypes)
    print("\nVacíos por columna:")
    print(df.isna().sum())
    print("\nFilas repetidas (idénticas):", df.duplicated().sum())
```

```python
creado = pd.to_datetime(tickets["created_at"], format="%Y-%m-%d")
resuelto = pd.to_datetime(tickets["resolved_at"], format="%Y-%m-%d", errors="coerce")
abiertos = tickets["resolved_at"].isna()
```

```python
#resueltos antes de crearse
print("\nResueltos antes de crearse:", (resuelto < creado).sum())
print("Resueltos después del 31/08:", (resuelto > "2025-08-31").sum())
```

```python
#CSAT (se entiende que es la calificacion de la resolucion del reclamo)
print("\nCSAT:", tickets["csat"].value_counts(dropna=False).sort_index().to_dict())
print("Abiertos con CSAT:", (abiertos & tickets["csat"].notna()).sum())
print("Cerrados sin CSAT:", (~abiertos & tickets["csat"].isna()).sum())

#no se entiendo la distribucion del CSAT (calificacion de la resolucion del reporte), algunos no estan cerrados pero ya tiene calificacion.
```

```python
#SLA (cumple el plazo maximo acordado para resolver el reclamo?)
print("\nSLA incumplido:", tickets["sla_breached"].value_counts().to_dict())
print("Abiertos con SLA incumplido:", (abiertos & tickets["sla_breached"]).sum())
dias = (resuelto - creado).dt.days
tickets.assign(dias=dias).groupby(["severity", "sla_breached"])["dias"].agg(["count", "mean", "min", "max"])
```
