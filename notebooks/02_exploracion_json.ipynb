```python
#rutas a los datos
LANDING = REPO / "datalake" / "landing"
EVENTS_DIR = LANDING / "usage_events_stream"
```

```python
#funcion para explorar
def explorar(df, nombre):
    print(f"=== {nombre} ===")
    print("Filas y columnas:", df.shape)
    print("\nTipo de cada columna:")
    print(df.dtypes)
    print("\nVacíos por columna:")
    print(df.isna().sum())
    print("\nFilas repetidas (idénticas):", df.duplicated().sum())
```

```python
archivos = sorted(EVENTS_DIR.glob("*.jsonl"))
print("Cantidad de archivos:", len(archivos))
print("Primero:", archivos[0].name, "| Último:", archivos[-1].name)

with open(archivos[0], encoding="utf-8") as f:
    for i in range(3):
        print(f.readline())
```

```python
import pandas as pd
```

```python
eventos = pd.concat(
    [pd.read_json(a, lines=True, dtype=False).assign(archivo=a.name) for a in archivos],
    ignore_index=True
)
explorar(eventos, "usage_events")
eventos.head(10)
```

```python
#duplicados en archivos json ditintos
print("event_id distintos:", eventos["event_id"].nunique(), "de", len(eventos))

#los id de eventos son unicos, no se repiten
```

```python
#campo schema_version
print("\nEventos por versión:", eventos["schema_version"].value_counts().to_dict())
print(eventos.groupby("schema_version")["timestamp"].agg(["min", "max"]))

#los datos del esquema 1 (schema_version) no traen el campo carbon_kg en el dato crudo, solo lo treae el esquema mas nuevo, el 2
```

```python
#tipo de dato de value
print("\nTipos en value:", eventos["value"].map(lambda v: type(v).__name__).value_counts().to_dict())
#tipo de dato distinto para value
```

```python
#vacios de metric y unit
print(pd.crosstab(eventos["metric"], eventos["unit"].fillna("VACÍO")))

#son solo 3 metricas, la unidad parece correcta en cada caso, el problema son los nulos en unit
```

```python
es_texto = eventos["value"].map(lambda v: isinstance(v, str))
textos = eventos.loc[es_texto, "value"]
print("Ejemplos:", textos.sample(10, random_state=1).tolist())
print("Textos que no se pueden pasar a número:", pd.to_numeric(textos, errors="coerce").isna().sum())
```

```python
#registros con value pero sin unit
print("\nCon value pero sin unit:", (eventos["value"].notna() & eventos["unit"].isna()).sum())
```

```python
#casos con montos negativos
print("\nCostos negativos:", (eventos["cost_usd_increment"] < 0).sum())
print(eventos["cost_usd_increment"].describe())

#2016 casos con montos negativos
```

```python
print(pd.crosstab(eventos["service"], eventos["metric"]))

print(eventos[eventos["genai_tokens"].notna()].groupby(["service", "schema_version"]).size())
```

```python
customers = pd.read_csv(LANDING / "customers_orgs.csv")
resources = pd.read_csv(LANDING / "resources.csv")

# Definir el conjunto de ids de clientes que faltaba
ids_clientes = set(customers["org_id"])
```

```python
#cruce con cutomer y resouces
print("\norg_id que no están en customers:", len(set(eventos["org_id"]) - ids_clientes))
print("resource_id que no están en resources:",
      len(set(eventos["resource_id"]) - set(resources["resource_id"])))
cruce_e = eventos.merge(resources[["resource_id", "org_id", "service"]],
                        on="resource_id", how="left", suffixes=("", "_res"))
print("Eventos cuyo cliente no es el dueño del recurso:", (cruce_e["org_id"] != cruce_e["org_id_res"]).sum())
print("Eventos con servicio distinto al del recurso:", (cruce_e["service"] != cruce_e["service_res"]).sum())
```

```python
#fecha del evento para comparar con las otras tablas
dia_evento = eventos["timestamp"].dt.strftime("%Y-%m-%d")

#antes del alta del cliente
alta = eventos["org_id"].map(customers.set_index("org_id")["signup_date"])
print("Eventos anteriores al alta del cliente:", (dia_evento < alta).sum())
```

```python
#antes de la creación del recurso
creacion = eventos["resource_id"].map(resources.set_index("resource_id")["created_at"])
print("Eventos anteriores a la creación del recurso:", (dia_evento < creacion).sum())
```

```python
#eventos por día, se distribuyen igual?
por_dia = dia_evento.value_counts()
print("Eventos por día, mínimo y máximo:", por_dia.min(), por_dia.max())
```
